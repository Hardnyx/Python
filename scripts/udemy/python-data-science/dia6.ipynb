```python
# 1. Cargar los datos

import pandas as pd
df = pd.read_csv('source/medallas.csv')
```

```python
# 2. Exploración Inicial

df.shape
df.columns
df.info()
df.head()
```

```python
# 3. Limpieza de Datos

df.isnull().sum()
df = df.fillna(0)
df[['Oro', 'Plata', 'Bronce']] = df[['Oro', 'Plata', 'Bronce']].astype(int)
df.head()
```

```python
# 4. Tres países con más medallas

top3 = df.sort_values(by='Total', ascending=False).head(3)
print(top3)
```

```python
min10medals = df[df['Total'] > 10]
min10medals
```
