```python
# 1. Lectura de datos
import pandas as pd
df1 = pd.read_csv('source/Datos_Ventas_Tienda.csv')
df2 = pd.read_csv('source/Datos_Ventas_Tienda2.csv')
```

```python
df2.head()
df2.shape
df2.info()
df2.describe()
```

```python
# 2. Fusión de datos
df = pd.merge(df1, df2, how = 'outer')
```
