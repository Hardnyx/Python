````python
import random as rnd
import re
from string import ascii_letters

WORD_BANK = [
    "BANANA", "BLIZZARD", "BISCUIT", "BUBBLE", "CACTUS", "CARROT", "CHOCOLATE",
    "DINOSAUR", "DONUT", "FLAMINGO", "GIGGLE", "GOOFY", "GIRAFFE", "JELLYBEAN",
    "LLAMA", "MARSHMALLOW", "MUFFIN", "NOODLE", "PANCAKE", "PENGUIN", "PICKLE",
    "PIRATE", "POPCORN", "POTATO", "PAJAMAS", "RUBBERDUCK", "SOCKS", "SLIPPER",
    "SNORKEL", "SPAGHETTI", "SQUIRREL", "TACO", "TICKLE", "TOASTER", "UNICORN",
    "WATERMELON", "WIGGLE", "WOBBLE", "ZOMBIE"
]

MAX_ATTEMPTS = 6

word_selected = rnd.choice(WORD_BANK)
word_user = '_'*len(word_selected)
word_on_screen = ('_ '*len(word_selected)).rstrip()


letters_guessed_right = []
letters_guessed_wrong = []

def is_ascii_letter(char:str):
    char = str(char)
    if not str(char).isalpha():
        print('You inserted something that is not a letter')
        return False
    if char in ascii_letters:
        if len(char) != 1:
            print('You inserted more than one letter')
            return False
    else:
        print('You inserted something that is not a standard letter')
        return False
        
    return True

def replace_char_in_string_by_index(string:str, char:str, index:int) -> str:
    string = string[:index] + char + string[index+1:]
    return string

def main():

    global actual_attempts
    global word_user
    global word_on_screen
    
    actual_attempts = MAX_ATTEMPTS

    print('A new word was generated!')
    print('Try to guess it!')
    print("Type 'close' to exit")
    print(f'{word_on_screen}\n')

    while actual_attempts > 0:
        
        user_letter = input('Insert a letter: ').upper()

        if user_letter == 'CLOSE':
            break

        if user_letter in letters_guessed_right:
            print('You already guessed that letter and was right')

        elif user_letter in letters_guessed_wrong:
            print('You already guessed that letter and was wrong')        

        elif is_ascii_letter(user_letter):

            if user_letter in word_selected:
                print('The letter is in the word')
                letters_guessed_right.append(user_letter)
                indexes = [m.start() for m in re.finditer(user_letter, word_selected)]

                for index in indexes:
                    word_user = replace_char_in_string_by_index(word_user, user_letter, index)
                    word_on_screen = replace_char_in_string_by_index(word_on_screen, user_letter, index*2)

                print(f'{word_on_screen}\n')
            
                if '_' not in word_user:
                    print('Congratulations, you won!')
                    break

            else:
                print('The letter is not in the word')
                letters_guessed_wrong.append(user_letter)
                actual_attempts -= 1
                print(f'Discarded letters: {letters_guessed_wrong}')
                
                if actual_attempts == 0: 
                    print(f'You have no attempts remaining\n')
                elif actual_attempts == 1:
                    print(f'You have {actual_attempts} attempt remaining\n')
                else:
                    print(f'You have {actual_attempts} attempts remaining\n') 

    else:
        print('You lost!')
        print(f'The word was {word_selected}')               

main()
````
