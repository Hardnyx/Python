# Problem 1

Write a Python program to read an entire text file.

```python
from pathlib import Path
def read_entire_file(filepath):
    with open(filepath, 'r') as f:
        content = f.read()
        print(content)

read_entire_file(Path('source') / 'problem1.txt')
```

# Problem 2

Write a Python program to read first n lines of a file.

```python
from pathlib import Path
def read_first_n_lines(filepath:int, n):
    with open(filepath) as f:
        content = ''
        for _ in range(n):
            line = f.readline()
            content = content + line
        print(content)

read_first_n_lines(Path('source') / 'problem1.txt', 2)
```

# Problem 3

Write a Python program to append text to a file and display the text. 

```python
from pathlib import Path
def append_text_to_file(filepath, text):
    with open(filepath, 'a+') as f:

        f.seek(0)
        content = f.read()
        if content:
            f.write('\n')
        
        f.write(text)
        f.seek(0)
        content = f.read()
        print(content)

append_text_to_file(Path('source') / 'problem4.txt', 'Hemos insertado una nueva línea')
```

# Problem 4

Write a Python program to read last n lines of a file.

```python
from pathlib import Path
def read_last_n_files(filepath, n):
    with open(filepath, 'r') as f:
        content = ''
        lines_number = sum(1 for _ in f)
        if n > lines_number:
            raise ValueError('n exceeds the number of lines of the file')

        f.seek(0)
        for _ in range(lines_number - n):
            f.readline()
        for _ in range(n):
            line = f.readline()
            content = content + line
        print(content)

read_last_n_files(Path('source') / 'problem4.txt', 2)
```

# Problem 5

Write a Python program to read a file line by line and store it into a list.

```python
def store_lines_in_list(pathfile):
    with open(pathfile, 'r') as f:
        lines_list = []
        for line in f:
            lines_list.append(line.rstrip())
    print(lines_list)
    return lines_list

store_lines_in_list(Path('source') / 'problem4.txt')
```
