# Problem 3

Write a NumPy program to test whether none of the elements of a given array are zero.

```python
import numpy as np
array_1 = np.array([1, 2, 3, 0])

def is_zero_in_array(ar):
    return np.any(ar == 0)

is_zero_in_array(array_1)
```
