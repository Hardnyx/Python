# Problem 1

Write a Python program to create a class representing a Circle. Include methods to calculate its area and perimeter.

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def perimeter(self):
        return 2 * math.pi * self.radius
    def area(self):
        return math.pi * self.radius ** 2

Circle(5).area()
```

# Problem 2

Write a Python program to create a person class. Include attributes like name, country and date of birth. Implement a method to determine the person's age.

```python
import datetime
from dateutil.relativedelta import relativedelta

class Person:
    def __init__(self, name:str, country:str, birthday:datetime.date):
        self.name = name
        self.country = country
        self.birthday = birthday
        
    def age(self):
        return relativedelta(datetime.date.today(), self.birthday).years

Person('Juan', 'Perú', datetime.date(2003,12,14)).age()
```

# Problem 4

Write a Python program to create a class that represents a shape. Include methods to calculate its area and perimeter. Implement subclasses for different shapes like circle, triangle, and square.

```python
import math

class RegularShape:
    def __init__(self, length:float, sides:int):
        self.length = length
        self.sides = sides

        if self.sides < 3:
            raise ValueError('A regular polygon must have at least 3 sides')
        
        if length <= 0:
            raise ValueError('A length value must be positive')

    def perimeter(self):
        return self.length * self.sides

    def area(self):
        return (self.sides * self.length**2) / (4 * math.tan(math.pi / self.sides))


class Circle(RegularShape):
    def __init__(self, length):
        super().__init__(length, 1)

    def perimeter(self):
        return 2 * math.pi * self.length
    
    def area(self):
        return math.pi * self.length ** 2
    

class Triangle(RegularShape):
    def __init__(self, length):
        super().__init__(length, 3)


class Square(RegularShape):
    def __init__(self, length):
        super().__init__(length, 4)
  

class Pentagon(RegularShape):
    def __init__(self, length):
        super().__init__(length, 5)

RegularShape(4, 3).perimeter()
```

```python
def tabber(text:str, left_tabs:int = 0, right_tabs:int = 0) -> str:
    new_text = text
    if left_tabs < 0 or right_tabs < 0:
        raise ValueError('Expected non negative value')
    else:
        return '\t' * left_tabs + text + '\t' * right_tabs

# Binary Search Tree

class Node():
    def __init__(self, node_number:int, hierarchy:int=None, parent=None, left_child=None, right_child=None):
        self.node_number = node_number
        self.parent = parent
        self.left_child = left_child
        self.right_child = right_child
        self.hierarchy = hierarchy


class BinarySearchTree:
    def __init__(self):
        self.root = None
        self.node_dict: dict[int, Node] = {}
        self.hierarchies_dict: dict[int, list[Node]] = {}

    def add_to_hierarchies_dict(self, current_node:Node):
        node_hierarchy = current_node.hierarchy
        if node_hierarchy not in self.hierarchies_dict:
            self.hierarchies_dict[node_hierarchy] = [current_node]
        else:
            self.hierarchies_dict[node_hierarchy].append(current_node)

    def _create_child_node(self, new_node:Node, parent_node:Node, node_number:int) -> Node:
        new_node = Node(node_number)
        new_node.parent = parent_node
        new_node.hierarchy = parent_node.hierarchy + 1

        self.node_dict[node_number] = new_node
        self.add_to_hierarchies_dict(new_node)

        return new_node

    def create_node(self, node_number:int) -> Node:

        if node_number in self.node_dict:
            raise ValueError('Node already exists')

        if len(self.node_dict) == 0:
            self.root = Node(node_number)
            self.root.hierarchy = 0
            self.node_dict[node_number] = self.root
            self.hierarchies_dict[0] = [self.root]
                    
        else:
            current_node = self.root
            while True:
                if node_number < current_node.node_number:
                    if current_node.left_child is None:
                        current_node.left_child = self._create_child_node(current_node.left_child, current_node, node_number)
                        break
                    else:
                        current_node = current_node.left_child
                else:
                    if current_node.right_child is None:
                        current_node.right_child = self._create_child_node(current_node.right_child, current_node, node_number)
                        break
                    else:
                        current_node = current_node.right_child
            
    def search_node_beta(self, node_number):

        if node_number not in self.node_dict:
            raise ValueError('Node does not exist in the tree')

        search_path = []
        node_searched = self.node_dict[node_number]
        search_path.insert(0, node_searched.node_number)

        while node_searched.parent is not None:
            node_searched = node_searched.parent
            search_path.insert(0, node_searched.node_number)
            
        print (f'The search path is {search_path}')
        return search_path

    def search_node(self, searched_node):
        
        if searched_node not in self.node_dict:
            raise ValueError('Node does not exist in the tree')

        current_node = self.root
        search_path = []
        search_path.append(current_node.node_number)

        while searched_node != current_node.node_number:
            if searched_node > current_node.node_number:
                current_node = current_node.right_child
                search_path.append(current_node.node_number)
            elif searched_node < current_node.node_number:
                current_node = current_node.left_child
                search_path.append(current_node.node_number)
            else:
                pass
        
        return search_path


    def print_tree_beta(self):
        line = ''
        for h in range(max(self.hierarchies_dict.keys()) + 1):
            for node in self.node_dict.values():
                if node.hierarchy == h:
                    line = line + ' ' + str(node.node_number)
            print(f'{line}\n')
            line = ''

    def print_tree_alpha(self):
        nodes_print_map: dict[int, str] = {}
        for h in range(max(self.hierarchies_dict.keys()) + 1):
            for n in self.hierarchies_dict[h]:
                
                text_line0 = ' ' + str(n.node_number) + ' '

                if n.left_child == None and n.right_child == None:
                    text_line1 = ''
                elif n.left_child != None and n.right_child == None:
                    text_line1 = '/ '
                elif n.left_child == None and n.right_child != None:
                    text_line1 = ' \\'
                else:
                    text_line1 = '/ \\'

                if n.parent != None:
                    if n == n.parent.left_child:
                        pass
                    elif n == n.parent.right_child:
                        text_line0 = tabber(text_line0, 1)
                        text_line1 = tabber(text_line1, 1)

                text = text_line0 + '\n' + text_line1
                nodes_print_map[n.node_number] = text
                print(nodes_print_map[n.node_number])


    def print_tree(self):
        nodes_print_map: dict[int, str] = {}
        global_text = ''

        for h in reversed(range(max(self.hierarchies_dict.keys()) + 1)):            
            line_nodes = ''
            line_branches = ''
            for n in self.hierarchies_dict[h]:
                line_nodes = line_nodes + str(n.node_number) + ' ' * (max(self.hierarchies_dict.keys()) + 1 - h)
                if n.parent == None:
                    break
                elif n.parent.left_child == n and n.parent.left_child is not None:
                    line_branches = line_branches + '/ ' + ' ' * (max(self.hierarchies_dict.keys()) + 1 - h) + '\\ ' + ' ' + ' ' * (max(self.hierarchies_dict.keys()) + 1 - h)
                elif n.parent.right_child == n and n.parent.left_child is not None:
                    continue
                elif n.parent.left_child == n:
                    line_branches = line_branches + '/ ' + ' ' * (max(self.hierarchies_dict.keys()) + 1 - h)
                elif n.parent.right_child == n:
                    line_branches = line_branches + '\\ ' + ' ' * (max(self.hierarchies_dict.keys()) + 1 - h)
            line_nodes = line_nodes.rstrip()
            line_branches = line_branches.rstrip()
            line_nodes = ' ' * 2 * (max(self.hierarchies_dict.keys()) - h) + line_nodes
            line_branches = ' ' + ' ' * 2 * (max(self.hierarchies_dict.keys()) - h) + line_branches
            #(max(self.hierarchies_dict.keys()) - h)

            global_text = global_text + line_nodes + '\n' + line_branches + '\n'
            
        global_text = "\n".join(global_text.splitlines()[::-1])
        print(global_text)

    #def delete_node(self, node_number):

    #    if self.node_dict[node_number].left_child is not None or self.node_dict[node_number].right_child is not None:
    #        raise ValueError('Can not delete a node with childs')
        
    #    else:
    #        del self.node_dict[node_number]



BST = BinarySearchTree()

for value in [50, 30, 70, 20, 40, 60, 80, 10, 25, 35, 45, 55, 65, 75, 90]:
    BST.create_node(value)

BST.print_tree()
```
