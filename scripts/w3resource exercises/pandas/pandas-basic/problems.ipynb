# Problem 2

Write a Pandas program to convert a Panda module Series to Python list and it's type.

# Problem 3

Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

```python
import pandas as pd
serie_1 = pd.Series([2, 4, 6, 8, 10])
serie_2 = pd.Series([1, 3, 5, 7, 9])

sum = serie_1 + serie_2
difference = serie_1 - serie_2
product = serie_1 * serie_2
division = serie_1 / serie_2

print(sum)
print(difference)
print(product)
print(division)
```

# Problem 4

Write a Pandas program to compare the elements of the two Pandas Series.

Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 10]

```python
import pandas as pd
serie_1 = pd.Series([2, 4, 6, 8, 10])
serie_2 = pd.Series([1, 3, 5, 7, 9])
result = serie_1 > serie_2
print(result)
```

# Problem 5

Sample Series: 
Original dictionary:
{'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 800}

Converted series:

a    100

b    200

c    300

d    400

e    800

dtype: int64 

```python
import pandas as pd
dict = {'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 800}
serie = pd.Series(dict.values(), dict.keys())
print(serie)
```
